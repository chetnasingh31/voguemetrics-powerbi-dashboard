```python
import pandas as pd
```

```python
df = pd.read_csv("/content/Myntra Fasion Clothing.csv")
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

```python
df.drop_duplicates()
```

```python
df.sort_values("discountprice_(in_rs)", ascending = False).head(10)
```

```python
df["discountprice_(in_rs)"].mean()
```

```python
df["category"].value_counts()
```

```python
df["discount_amount"] = (
    df["originalprice_(in_rs)"] -
    df["discountprice_(in_rs)"]
)
```

```python
df[[
    "originalprice_(in_rs)",
    "discountprice_(in_rs)",
    "discount_amount"
]].head()
```

```python
df["discount_percent"] = (
    df["discount_amount"] /
    df["originalprice_(in_rs)"]
) * 100
```

```python
df[["discount_percent","discount_amount","originalprice_(in_rs)"]].head(5)
```

```python
import matplotlib.pyplot as plt

df["category"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Product Categories")
plt.xlabel("Category")
plt.ylabel("Count")

plt.show()
```

```python
df["ratings"].describe()
```

```python
import matplotlib.pyplot as plt

df["ratings"].plot(kind="hist", bins=20)

plt.title("Ratings Distribution")
plt.xlabel("Ratings")
plt.ylabel("Frequency")

plt.show()
```

```python
df.groupby("brandname")["ratings"].mean().sort_values(ascending=False).head(10)
```

```python
df[["ratings", "reviews"]].corr()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["reviews"], df["ratings"])

plt.title("Reviews vs Ratings")
plt.xlabel("Reviews")
plt.ylabel("Ratings")

plt.show()
```

```python
df["category_by_gender"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["category_by_gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Product Distribution by Gender")
plt.ylabel("")

plt.show()
```

```python
df.groupby("brandname")["originalprice_(in_rs)"] \
.mean() \
.sort_values(ascending=False) \
.head(10)
```

```python
df.groupby("brandname")["originalprice_(in_rs)"] \
.mean() \
.sort_values(ascending=False) \
.head(10) \
.plot(kind="bar")

plt.title("Top 10 Premium Brands")
plt.xlabel("Brand")
plt.ylabel("Average Original Price")

plt.show()
```

```python
df.groupby("brandname")["discount_percent"] \
.mean() \
.sort_values(ascending=False) \
.head(10) \
.plot(kind="bar")

plt.title("Top 10 Brands by Average Discount")
plt.xlabel("Brand")
plt.ylabel("Average Discount %")

plt.show()
```
